#### Introducing the SentenceTransformer Encoding LLM

The all-MiniLM is a very useful model from HuggingFace that maps sentences & paragraphs to 384 dimensional vectors and is ideal for tasks like semantic search.

https://huggingface.co/sentence-transformers/all-MiniLM-L6-v2

It can run pretty quickly locally.

As an alternative, OpenAI provides a closed-source Embeddings model. Benefits compared to OpenAI embeddings:

1. It's free and fast!
2. We can run it locally, so the data never leaves our box - might be useful if you're building a personal RAG

````python
encoder = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
````

````python
# Pass in a list of texts, get back a numpy array of vectors

vector = encoder.encode(["A proficient AI engineer who has almost reached the finale of AI Engineering Core Track!"])[0]
print(vector.shape)
````

##### With that background, let's populate our Chroma database


By calculating vectors for 800,000 scraped products

This takes 30 minutes on my machine on my GPU - it might take longer for you - feel free to use the Lite dataset!

````python
# Check if the collection exists; if not, create it

collection_name = "products"
existing_collection_names = [collection.name for collection in client.list_collections()]

if collection_name not in existing_collection_names:
    collection = client.create_collection(collection_name)
    for i in tqdm(range(0, len(train), 1000)):
        documents = [item.summary for item in train[i: i+1000]]
        vectors = encoder.encode(documents).astype(float).tolist()
        metadatas = [{"category": item.category, "price": item.price} for item in train[i: i+1000]]
        ids = [f"doc_{j}" for j in range(i, i+1000)]
        ids = ids[:len(documents)]
        collection.add(ids=ids, documents=documents, embeddings=vectors, metadatas=metadatas)

collection = client.get_or_create_collection(collection_name)
````

````python
def vector(item):
    return encoder.encode(item.summary)
````

````python
def find_similars(item):
    vec = vector(item)
    results = collection.query(query_embeddings=vec.astype(float).tolist(), n_results=5)
    documents = results['documents'][0][:]
    prices = [m['price'] for m in results['metadatas'][0][:]]
    return documents, prices
````

````python
print(find_similars(test[0]))
````

````python
# We need to give some context to GPT-5.1 by selecting 5 products with similar descriptions

def make_context(similars, prices):
    message = "For context, here are some other items that might be similar to the item you need to estimate.\n\n"
    for similar, price in zip(similars, prices):
        message += f"Potentially related product:\n{similar}\nPrice is ${price:.2f}\n\n"
    return message
````

````python
documents, prices = find_similars(test[0])
print(make_context(documents, prices))
````

````python
def messages_for(item, similars, prices):
    message = f"Estimate the price of this product. Respond with the price, no explanation\n\n{item.summary}\n\n"
    message += make_context(similars, prices)
    return [{"role": "user", "content": message}]
````

````python
documents, prices = find_similars(test[0])
print(messages_for(test[0], documents, prices)[0]['content'])
````

#### Deploying Open-Weight Models in a Modal Endpoint

Follow the steps below to deploy an open-weight model using a dedicated endpoint on [Modal](https://modal.com/).

##### 1. Log in to Modal

Log in to your Modal account and navigate to **Endpoints**.

##### 2. Create a New Endpoint

Click **Create a New Endpoint** and select:

> **Deploy a dedicated endpoint — Billed by GPU second**

This option allows you to deploy the model on dedicated GPU infrastructure and pay based on GPU usage.

##### 3. Select a Model

Select the model you want to deploy.

For this setup, we are using:

```text
Qwen 3.5 35B A3B FP8
```

This model is a good choice for **agent-based workloads**.

##### 4. Configure the Endpoint

Give your endpoint a meaningful name and configure the required authentication.

While creating the endpoint, set the **Container Limit** to:

```text
300 seconds
```

If the endpoint remains inactive for **300 seconds**, it will be placed into an **IDLE state**.

This helps reduce GPU costs when the endpoint is not actively being used.

##### 5. Get the Endpoint Credentials

Once the endpoint has been created, you will have access to:

* ```Endpoint URL```
* ```MODAL_PROXY_TOKEN_ID```
* ```MODAL_PROXY_TOKEN_SECRET```

These credentials are required to authenticate requests to your Modal endpoint.

##### 6. Configure Environment Variables

Set the following environment variables in your ```.env``` file or ```.bashrc```:

```
MODAL_ENDPOINT_URL="your_modal_endpoint_base_url"
MODAL_PROXY_TOKEN_ID="your_token_id"
MODAL_PROXY_TOKEN_SECRET="your_token_secret"
```

> **Note:** Keep `MODAL_PROXY_TOKEN_SECRET` private.
> Do not commit it to source control or expose it in publicly shared notebooks.

##### 7. Verify the Environment Variables

You can verify that the variables are configured correctly:

```
echo $MODAL_ENDPOINT_URL
echo $MODAL_PROXY_TOKEN_ID
```

You can now use these environment variables to authenticate and send requests to your deployed open-weight model through the Modal endpoint.

````python
# This function will call the qwen model in Modal.com

def qwen_rag_modal(item):
    import litellm
    litellm.api_base = (os.environ['MODAL_ENDPOINT_URL'])
    litellm.api_key = (
        f"{os.environ['MODAL_PROXY_TOKEN_ID']}."
        f"{os.environ['MODAL_PROXY_TOKEN_SECRET']}"
    )
    response = litellm.completion(
        model="openai/Qwen/Qwen3.5-35B-A3B-FP8",
        messages=messages_for(item, documents, prices),
        temperature=0.7,
        max_tokens=2048,
        top_p=0.9,
        stream=False,
        extra_body={"reasoning_effort": "none"},
        seed=42,
    )
    # print(response)
    return response.choices[0].message.content
````

````python
import modal
Pricer = modal.Cls.from_name("pricer-service", "Pricer")
pricer = Pricer()
````

````python
def specialist(item):
    return pricer.price.remote(item.summary)
````

````python
def get_price(reply):
    reply = reply.replace("$", "").replace(",", "")
    match = re.search(r"[-+]?\d*\.\d+|\d+", reply)
    return float(match.group()) if match else 0
````

````python
def ensemble(item):
    price1 = get_price(qwen_rag_modal(item))
    price2 = specialist(item)
    price3 = deep_neural_network(item)

    return price1*0.8 + price2*0.1 + price3*0.1
````
